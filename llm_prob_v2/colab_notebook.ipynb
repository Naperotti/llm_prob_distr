## 3. Start Server

Starting the FastAPI server. Once running, look for the URL that Colab displays above this cell.

```python
import asyncio
import nest_asyncio
import uvicorn
from uvicorn import Config, Server

# Allow nested event loops in Colab
nest_asyncio.apply()

print("✅ Starting server on port 8000...")
print("👉 After the server starts, Colab will show a URL link above this cell")
print("   Click that link to open the app\n")

# Create and run server in existing event loop
config = Config(app="app:app", host="0.0.0.0", port=8000, reload=False, log_level="info")
server = Server(config)
await server.serve()
```

## Usage

Once the server is running:

1. **Look above the running cell** - Colab will show a clickable URL (looks like a link icon)
2. **Click that URL** to open the app in a new tab
3. **Generate sequences:**
   - Enter a prompt (e.g., "The future of AI is")
   - Adjust settings if desired
   - Click "Generate Sequences"
4. **Analyze with UMAP:**
   - After generating, click "Analyze UMAP"
   - View the 2D semantic similarity plot
5. **Save/Load datasets:**
   - Click "Save Dataset" to store generated data
   - Click "Load Dataset" to restore previous data

## Notes

- **GPU acceleration**: The code automatically detects and uses Colab's GPU (T4)
- Generation is **much faster** on GPU (~2-3 seconds vs ~10-20 seconds on CPU)
- First generation will be slower (~10s) as the model loads to GPU
- The Colab URL only works while you're logged in to your Google account
- Datasets are saved in Colab's temporary storage (lost when runtime stops)
- Model: Qwen2.5-1.5B (1.5 billion parameters, better than GPT-2)

## Troubleshooting

- **"Can't find the URL"**: Look for a small link icon above the running cell
- **"Application startup failed"**: Restart the runtime and try again
- **Slow generation**: Make sure you enabled GPU (Runtime → Change runtime type)
- **Out of memory**: Reduce num_sequences or num_tokens in the UI
