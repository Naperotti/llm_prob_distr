## 📖 How to Use

### Quick Start:
1. **Open the interface** using the link above (right-click → Open in New Tab)
2. **Enter a prompt** in the text area
3. **Adjust parameters**: top-k, top-p, temperature, min-p
4. **Click "Get probs"** to see token distribution
5. **Click tokens** to extend the sequence

### Branching Visualization:
1. Set **N** (number of sequences, try 50-500)
2. Set **Max Length** (tokens per sequence, try 20-50)
3. Click **"Generate"** in the Branching section
4. See where sequences diverge (branching tree)
5. Explore semantic clustering (UMAP scatter plot)

### UMAP Parameters:
- **n_neighbors** (2-50): Lower = tight clusters, Higher = global structure
- **min_dist** (0-1): 0 = tight packing, 0.5 = more spread
- **spread** (0.1-3): Controls overall scale of points

### Tips:
- ✅ GPU acceleration makes 100-500 sequences fast!
- ✅ Hover over tokens to see probability distributions
- ✅ Click UMAP points to see full sequence text
- ✅ Lower temperature (0.3-0.7) = more diverse branches
- ✅ All tooltips explain what each parameter does

### Troubleshooting:
- **Interface not loading?** Wait 15 seconds and refresh
- **API errors?** Check that server started successfully (see Cell 3 output)
- **Slow generation?** Check GPU is enabled (Cell 1)
- **Need to restart?** Interrupt execution and re-run Cell 3
