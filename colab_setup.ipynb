```python
# Start server with ngrok tunnel
from pyngrok import ngrok
import nest_asyncio
import os

nest_asyncio.apply()

# Optional: Set ngrok auth token (get free token at https://dashboard.ngrok.com/get-started/your-authtoken)
# Uncomment and add your token if you want longer session times:
# ngrok.set_auth_token("YOUR_TOKEN_HERE")

# Start ngrok tunnel
public_url = ngrok.connect(8000)
print(f"\n🌐 Your app is running at: {public_url}")
print(f"\n📝 Open this URL in your browser to use the app!\n")

# Modify the index.html to use this URL automatically
import re
with open('index.html', 'r', encoding='utf-8') as f:
    html = f.read()

# Replace localhost with ngrok URL
html = re.sub(
    r'value="http://localhost:8000"',
    f'value="{public_url}"',
    html
)

# Save modified HTML
with open('index_colab.html', 'w', encoding='utf-8') as f:
    f.write(html)

print("✅ Created index_colab.html with auto-configured API URL")
print("   Download it from the Files panel (📁) on the left and open in browser\n")

# Start the FastAPI server
print("🚀 Starting FastAPI server...\n")
!python -m uvicorn app:app --host 0.0.0.0 --port 8000
```
